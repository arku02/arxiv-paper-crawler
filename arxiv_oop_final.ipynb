```python
# 建立 主table

with connection.cursor() as cursor :
    sql = """
        CREATE TABLE IF NOT EXISTS papers(
        id  INT AUTO_INCREMENT PRIMARY KEY,
        title VARCHAR(500),
        summary TEXT,
        published DATE,
        link VARCHAR(500) UNIQUE,
        created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
        )
    """
    cursor.execute(sql)
    cursor.execute("DESC papers")
    result = cursor.fetchall()
    for row in result:
        print(row)
        
print(f"建立table papers")
```

```python
# 因為一篇論文通常作者有好幾個
# 所以建一個 作者表

with connection.cursor() as cursor :
    sql = """
        CREATE TABLE IF NOT EXISTS authors(
            id INT AUTO_INCREMENT PRIMARY KEY,
            paper_id INT,
            name VARCHAR(255),
            FOREIGN KEY (paper_id) REFERENCES papers(id)
        )
    """
    cursor.execute(sql)
    cursor.execute("DESC authors")
    result = cursor.fetchall()
    for row in result:
        print(row)
    

print(f"建立table authers")
```

```python
# 紀錄 爬取的紀錄表 search_logs

with connection.cursor() as cursor :
    sql = """
        CREATE TABLE IF NOT EXISTS search_logs(
            id INT AUTO_INCREMENT PRIMARY KEY,
            query VARCHAR(255),
            result_count INT,
            search_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
        )
    """
    cursor.execute(sql)

print(f"建立search_logs")
```

```python
# 寫入 papers 和 authors

new_count = 0 
# 成功新增的

duplicate_count = 0 
# 被阻止寫入的重複筆數

with connection.cursor() as cursor :
    for paper in papers:
        try:
            sql = """
                INSERT INTO papers(title, summary, published, link)
                VALUES (%s, %s, %s, %s)
            """
            cursor.execute(sql, (
                paper["論文名"],
                paper["摘要"],
                paper["發表時間"],
                paper["論文網址"]
            ))

            paper_id = cursor.lastrowid

            sql = """
                    INSERT INTO authors(paper_id, name)
                    VALUES (%s, %s)
                """

            for author in paper["作者"]:
                cursor.execute(sql, (paper_id, author))
            
            new_count += 1
        
        except pymysql.err.IntegrityError:
            duplicate_count += 1
    
    sql = """
        INSERT INTO search_logs (query, result_count)
        VALUES (%s, %s)
    """

    cursor.execute(sql, ("cat:cs.CV", new_count))

connection.commit()
print(f"新增 {new_count} 筆, 略過 {duplicate_count} 筆")
```

## 物件導向版本（完成版）

把前面所有步驟封裝成一個 `Arxiv` 類別：搜尋 → 整理 → 存 CSV / MySQL → 查詢。

```python
# ============================================================
# 物件導向版 arXiv 爬蟲（完成版）
# ============================================================
import requests
import pandas as pd
import pymysql
from bs4 import BeautifulSoup as bs
from configparser import ConfigParser


class Arxiv:
    """arXiv 論文爬蟲：搜尋 → 整理 → 存 CSV / MySQL → 查詢"""

    def __init__(self, max_results=20):
        self.url = "http://export.arxiv.org/api/query"
        self.max_results = max_results
        self.papers = []
        self.query = ""
        self.db = None          # 先給預設值，沒連線就呼叫也不會 AttributeError

    # ---------------- 爬取 ----------------
    def search(self, query):
        self.query = query
        params = {
            'search_query': query,
            'start': 0,
            'max_results': self.max_results,
            'sortBy': 'submittedDate',
            'sortOrder': 'descending'
        }

        res = requests.get(self.url, params=params)

        if res.status_code != 200:
            print(f"狀態碼： {res.status_code}")
            return

        self.papers = self._parse(res.text)
        print(f"取得 {len(self.papers)} 篇論文")

    def _parse(self, xml_text):
        soup = bs(xml_text, "xml")
        entries = soup.find_all("entry")

        papers = []
        for entry in entries:
            authors_list = [a.find("name").text for a in entry.find_all("author")]

            papers.append({
                "論文名": entry.find("title").text.strip(),
                "摘要": entry.find("summary").text.strip(),
                "發表時間": entry.find("published").text[:10],
                "作者": authors_list,
                "論文網址": entry.find("id").text
            })

        return papers

    # ---------------- 呈現 / 匯出 ----------------
    def paper_dataFrame(self):
        if not self.papers:
            print("沒有資料")
            return
        return pd.DataFrame(self.papers)

    def show_paper(self, index=0):
        if not self.papers:
            print("沒有資料")
            return

        paper = self.papers[index]
        print(f"論文標題 : {paper['論文名']}")
        print(f"發表時間 : {paper['發表時間']}")
        print(f"作者     : {', '.join(paper['作者'])}")
        print(f"論文摘要 : {paper['摘要']}")
        print(f"論文網址 : {paper['論文網址']}")

    def save_csv(self, filename="cs_papers.csv"):
        if not self.papers:
            print("沒有資料")
            return

        pd.DataFrame(self.papers).to_csv(filename, index=False, encoding="utf-8-sig")
        print(f"儲存了 {len(self.papers)} 筆論文到 {filename}")

    # ---------------- 資料庫 ----------------
    def connect_db(self, config_file="config.ini", database="arxiv_db"):
        config = ConfigParser()
        config.read(config_file, encoding="utf-8")

        self.db = pymysql.connect(
            host=config.get('DB', 'host'),
            user=config.get('DB', 'user'),
            password=config.get('DB', 'password'),
            port=config.getint('DB', 'port'),
            charset="utf8mb4",
            cursorclass=pymysql.cursors.DictCursor
        )

        with self.db.cursor() as cursor:
            cursor.execute(f"CREATE DATABASE IF NOT EXISTS {database}")
            cursor.execute(f"USE {database}")

            cursor.execute("""
                CREATE TABLE IF NOT EXISTS papers(
                    id         INT AUTO_INCREMENT PRIMARY KEY,
                    title      VARCHAR(500),
                    summary    TEXT,
                    published  DATE,
                    link       VARCHAR(500) UNIQUE,
                    created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
                )
            """)

            cursor.execute("""
                CREATE TABLE IF NOT EXISTS authors(
                    id       INT AUTO_INCREMENT PRIMARY KEY,
                    paper_id INT,
                    name     VARCHAR(255),
                    FOREIGN KEY (paper_id) REFERENCES papers(id)
                )
            """)

            cursor.execute("""
                CREATE TABLE IF NOT EXISTS search_logs(
                    id           INT AUTO_INCREMENT PRIMARY KEY,
                    query        VARCHAR(255),
                    result_count INT,
                    searched_at  TIMESTAMP DEFAULT CURRENT_TIMESTAMP
                )
            """)

        self.db.commit()
        print("資料庫已連線，papers / authors / search_logs 三張表建立完成")

    def save_db(self):
        if not self.papers:
            print("沒有資料")
            return
        if not self.db:
            print("沒有連線到資料庫，請先呼叫 connect_db()")
            return

        new_count = 0
        dup_count = 0

        with self.db.cursor() as cursor:
            for paper in self.papers:
                try:
                    cursor.execute("""
                        INSERT INTO papers(title, summary, published, link)
                        VALUES (%s, %s, %s, %s)
                    """, (
                        paper["論文名"],
                        paper["摘要"],
                        paper["發表時間"],
                        paper["論文網址"]
                    ))

                    paper_id = cursor.lastrowid

                    for author in paper["作者"]:
                        cursor.execute("""
                            INSERT INTO authors(paper_id, name)
                            VALUES (%s, %s)
                        """, (paper_id, author))

                    new_count += 1

                except pymysql.err.IntegrityError:
                    dup_count += 1

            cursor.execute("""
                INSERT INTO search_logs (query, result_count)
                VALUES (%s, %s)
            """, (self.query, new_count))

        self.db.commit()
        print(f"新增 {new_count} 筆，略過重複 {dup_count} 筆")

    # ---------------- 查詢 ----------------
    def query_papers(self, keyword=None):
        if not self.db:
            print("沒有連線到資料庫，請先呼叫 connect_db()")
            return

        with self.db.cursor() as cursor:
            if keyword:
                cursor.execute("""
                    SELECT id, title, published, link
                    FROM papers
                    WHERE title LIKE %s OR summary LIKE %s
                    ORDER BY published DESC
                """, (f"%{keyword}%", f"%{keyword}%"))
            else:
                cursor.execute("""
                    SELECT id, title, published, link
                    FROM papers
                    ORDER BY published DESC
                """)
            results = cursor.fetchall()

        print(f"查到 {len(results)} 筆論文")
        return pd.DataFrame(results)

    def query_authors(self, paper_id):
        if not self.db:
            print("沒有連線到資料庫，請先呼叫 connect_db()")
            return

        with self.db.cursor() as cursor:
            cursor.execute("""
                SELECT p.title AS 論文名, a.name AS 作者
                FROM papers p
                JOIN authors a ON p.id = a.paper_id
                WHERE p.id = %s
            """, (paper_id,))
            results = cursor.fetchall()

        return pd.DataFrame(results)

    def show_logs(self):
        if not self.db:
            print("沒有連線到資料庫，請先呼叫 connect_db()")
            return

        with self.db.cursor() as cursor:
            cursor.execute("""
                SELECT query, result_count, searched_at
                FROM search_logs
                ORDER BY searched_at DESC
            """)
            results = cursor.fetchall()

        df = pd.DataFrame(results)
        print(f"總共有 {len(df)} 筆爬取紀錄")
        return df

    def close_db(self):
        if self.db:
            self.db.close()
            self.db = None
            print("資料庫連線已關閉")
        else:
            print("目前沒有開啟中的連線")
```
