```python
"""
http://export.arxiv.org/api/query?search_query=cat:cs.CV&max_results=2 
arxiv 指定的API基本網址用法
cs.CV -> cs = Cumputer science, CV = Computer Vision

ti -> Title, au -> Author, abs -> Abstract, co -> Comment
jr -> Journal Reference, cat -> Subject Category
rn -> Report Number, id -> Id (use id_list instead)
all -> All of the above

?xml version='1.0' -> xml的版本

<feed 
    xmlns:opensearch="http://a9.com/-/spec/opensearch/1.1/" 
    xmlns:arxiv="http://arxiv.org/schemas/atom" 
    xmlns="http://www.w3.org/2005/Atom"
    >

xmlns = XML Name Space
xmlns="http://www.w3.org/2005/Atom" -> 預設命名
例如: title, summary, author, id, published

xmlns:opensearch="http://a9.com/-/spec/opensearch/1.1/"
-> 有 opensearch: 都是屬於這種 
    <opensearch:itemsPerPage>2</opensearch:itemsPerPage>
    <opensearch:totalResults>192035</opensearch:totalResults>
    <opensearch:startIndex>0</opensearch:startIndex>

xmlns:arxiv="http://arxiv.org/schemas/atom" 
-> 有 arxiv: 都是arxiv 自己定義的
    <arxiv:primary_category term="cs.CV"/>

"""
```

```python
soup = bs(res.text, "xml")
#print(soup)
print(soup.title)
print(soup.title.text)
```

```python
entries = soup.find_all("entry")
for entry in entries:
    print(entry)

print(entries)
print(f"找到{len(entries)}個entries")
```

```python
# title
entry_1 = entries[0]
print(entry_1)
entry_1_title = entry_1.find("title").text
print(entry_1_title)
print(repr(entry_1_title))
# 去除前後空白和換行
entry_1_title_strip = entry_1_title.strip()
print(entry_1_title_strip)
```

```python
# summary

summary = entry_1.find("summary").text.strip()

print(summary)
```

```python
# date
published = entry_1.find("published").text
print(f"原本的日期:{published}")
published_main = published[:10]
print(f"需要的日期:{published_main}")
```

```python
# 作者(通常多個)
authors = entry_1.find_all("author")
print(f"總共有{len(authors)}個作者")

authors_list = []
for a in authors:
    author_name = a.find("name")
    authors_list.append(author_name.text)

print(authors_list)
print(f"論文作者列:{authors_list}")
```

```python
# 論文的網址,id
link = entry_1.find("id").text
print(f"論文連結：{link}")
```

```python
# 整理
paper = {
    "論文名" : entry_1_title_strip,
    "摘要" : summary,
    "發表時間" : published_main,
    "作者" : authors_list,
    "論文網址" : link
}
for key, value in paper.items():
    print(f"\n{key}")
    print(value)
```

```python
# 全部整理
papers = []
for entry in entries:
    paper = {
        "論文名" : entry_1_title_strip,
        "摘要" : summary,
        "發表時間" : published_main,
        "作者" : authors_list,
        "論文網址" : link
    }
    papers.append(paper)

print(f"總共{len(papers)}篇論文")
print(f"第一篇論文名:{papers[0]["論文名"]}")
print(f"第一篇論文名:{papers[0]["作者"]}")
print(f"第二篇論文名:{papers[1]["論文名"]}")
print(f"第二篇論文名:{papers[1]["作者"]}")

# 錯誤
```
